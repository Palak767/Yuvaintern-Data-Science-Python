```python
yearly_happiness = (
    df_clean
    .groupby("Year")["Life_Evaluation"]
    .mean()
    .reset_index()
)
yearly_happiness
```

```python
plt.figure(figsize=(12, 6))
plt.plot(
    yearly_happiness["Year"],
    yearly_happiness["Life_Evaluation"],
    marker="o"
)
plt.title("Average Life Evaluation Over Time")
plt.xlabel("Year")
plt.ylabel("Average Life Evaluation")
plt.xticks(yearly_happiness["Year"], rotation=45)
plt.tight_layout()
plt.savefig(
    "happiness_trend.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
yearly_happiness
highest_year = yearly_happiness.loc[
    yearly_happiness["Life_Evaluation"].idxmax()
]

lowest_year = yearly_happiness.loc[
    yearly_happiness["Life_Evaluation"].idxmin()
]

print("Highest average happiness year:")
print(highest_year)

print("\nLowest average happiness year:")
print(lowest_year)
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(
    factor_analysis["GDP_per_Capita"],
    factor_analysis["Life_Evaluation"],
    alpha=0.6
)
plt.title("GDP per Capita vs Life Evaluation")
plt.xlabel("Log GDP per Capita")
plt.ylabel("Life Evaluation")
plt.tight_layout()
plt.savefig(
    "gdp_vs_happiness.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
correlation_value = factor_analysis[
    "GDP_per_Capita"
].corr(
    factor_analysis["Life_Evaluation"]
)
print("Correlation between GDP and Life Evaluation:",
      round(correlation_value, 3))
```

```python
correlation_columns = [
    "Life_Evaluation",
    "GDP_per_Capita",
    "Social_Support",
    "Healthy_Life_Expectancy",
    "Freedom",
    "Generosity",
    "Corruption"
]

correlation_matrix = factor_analysis[
    correlation_columns
].corr()
plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Matrix of Happiness Factors")

plt.tight_layout()

plt.savefig(
    "correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
life_correlations = (
    correlation_matrix["Life_Evaluation"]
    .drop("Life_Evaluation")
    .sort_values(ascending=False)
)

life_correlations
```

```python
country_happiness = (
    df_clean
    .groupby("Country")["Life_Evaluation"]
    .mean()
    .sort_values(ascending=False)
)

top_10_countries = country_happiness.head(10)
top_10_countries

plt.figure(figsize=(10, 6))

plt.barh(
    top_10_countries.index[::-1],
    top_10_countries.values[::-1]
)

plt.title("Top 10 Countries by Average Life Evaluation")
plt.xlabel("Average Life Evaluation")
plt.ylabel("Country")

plt.tight_layout()

plt.savefig(
    "top_10_countries.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
print(
    "Country with highest average life evaluation:",
    country_happiness.idxmax()
)

print(
    "Average score:",
    round(country_happiness.max(), 3)
)

print(
    "Country with lowest average life evaluation:",
    country_happiness.idxmin()
)

print(
    "Average score:",
    round(country_happiness.min(), 3)
)
```
